```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cleaned_house_prices.csv")

df.head()
```

1)Distribution of deal contracts

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Amount_numeric"],
    bins=50,
    log_scale=True
)

plt.title("Distribution of House Prices")
plt.xlabel("Price (₹)")
plt.ylabel("Number of Properties")

plt.show()
```

2)Plot: Price vs Carpet Area

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Carpet_Area_sqft",
    y="Amount_numeric",
    alpha=0.4
)

plt.title("Price vs Carpet Area")
plt.xlabel("Carpet Area (sqft)")
plt.ylabel("Price (₹)")
plt.show()
```

3)Price Distribution by Transaction Type

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Transaction",
    y="Amount_numeric"
)

plt.title("House Prices by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Price (₹)")
plt.show()
```

4)Average Price by Location

```python
top_locations = (
    df.groupby("location")["Amount_numeric"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_locations.values,
    y=top_locations.index
)

plt.title("Top 10 Locations by Average House Price")
plt.xlabel("Average Price (₹)")
plt.ylabel("Location")
plt.show()
```

5)Price vs Number of Bathrooms

```python
bathroom_price = (
    df.groupby("Bathroom")["Amount_numeric"]
    .mean()
    .sort_index()
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=bathroom_price.index,
    y=bathroom_price.values
)

plt.title("Average House Price by Number of Bathrooms")
plt.xlabel("Number of Bathrooms")
plt.ylabel("Average Price (₹)")
plt.show()
```

6)Price by Furnishing Status

```python
furnishing_price = (
    df.groupby("Furnishing")["Amount_numeric"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=furnishing_price.index,
    y=furnishing_price.values
)

plt.title("Average House Price by Furnishing Status")
plt.xlabel("Furnishing Status")
plt.ylabel("Average Price (₹)")
plt.show()
```
